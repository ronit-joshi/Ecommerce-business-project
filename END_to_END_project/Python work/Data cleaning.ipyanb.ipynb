```python
# step 1: import libraries
import pandas as pd
import numpy as np
```

```python
# step 9: Remove duplicate values     
df.drop_duplicates(inplace=True)
```

```python
# Step 10: Again check duplicates
df.duplicated().sum()
```

```python
# Step 11: Replace invalid values
# Replace common invalid values with NaN 
df.replace(['N/A','Null',' '],np.nan,inplace=True)
```

```python
# step 12:Remove extra blank rows
df=df.dropna(how='all')
```

```python
# step 13:Remove extra blank columns
df=df.dropna(axis=1,how='all')
```

```python
# step 14:Remove extra spaces  of text/data values
df=df.map(lambda x: x.strip() if isinstance(x,str) else x)
```

```python
# step 15: Standardize text formet 
# Convert text into proper Case.
df['Customer_Name']=df['Customer_Name'].str.title()
df['City']=df['City'].str.title()
df['State']=df['State'].str.title()
df['Product']=df['Product'].str.title()
df['Category']=df['Category'].str.title()
df['Order_Status']=df['Order_Status'].str.title()
```

```python
# Step 16: Volidate email addresses
#  Keep only valid email records 
df=df[df['Email'].str.contains('@',na=False)]
```

```python
# step 17: Convert date columns
df['Order_Date']=pd.to_datetime(df['Order_Date'],dayfirst=True,errors='coerce')
df['Delivery_Date']=pd.to_datetime(df['Delivery_Date'],dayfirst=True,errors='coerce')
```

```python
# step 18: Convert Numeric Columns
cols=['Qty','Unit_Price','Discount']
for c in cols:
    df[c]=pd.to_numeric(df[c],errors='coerce')
```

```python
# step 19: Remove Invalid id Quantity.
# Remove records having negative or zeros quantity
df=df[df['Qty']>0]
```

```python
# step 20: fill missing values 
df['Discount']=df['Discount'].fillna(0)
df['Phone']=df['Phone'].fillna('unknowm')
df['Delivery_Date']=df['Delivery_Date'].fillna(df['Order_Date'])
```

```python
# step 21 frature engineering
#total sales
df['Sales']=df['Qty']*df['Unit_Price']
```

```python
df.head(2)
```

```python
# step 22: net amount
df['NET_Amount']=df['Sales']-(df['Sales']*df['Discount']/100)
```

```python
df.head(2)
```

```python
# step 23: profit(Example:20%)
df['Profit']=(df['NET_Amount']*0.20).round(2)
```

```python
df.head(2)
```

```python
#step 24:order month find by order date 
df['Month']=df['Order_Date'].dt.month_name()
```

```python
df.head(2)
```

```python
#step 25:order month find by order year
df['Year']=df['Order_Date'].dt.year
```

```python
#step 24:order month find by order weekday 
df['weak_day']=df['Order_Date'].dt.day_name()
```
